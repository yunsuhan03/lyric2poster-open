# 03. Explore the chart history (Big Data part)
The GitHub Action appends one row per track per day to `data/history/*.csv`. Here you analyze it with pandas.

```python
!git clone https://github.com/<your-user>/lyric2poster-open.git app 2>/dev/null || (cd app && git pull)
%cd app
import pandas as pd, glob
files = glob.glob('data/history/*.csv')
print(files)
df = pd.concat([pd.read_csv(f).assign(chart=f.split('/')[-1][:-4]) for f in files]) if files else pd.DataFrame()
df.head()
```

```python
# Which songs stayed in the top 10 the longest?
if not df.empty:
    top = df[df['rank'] <= 10].groupby(['chart', 'title', 'artist']).size().sort_values(ascending=False).head(15)
    display(top)
```
